```python
data = "/content/drive/MyDrive/TicketMind/data/Bitext_Sample_Customer_Support_Training_Dataset_27K_responses-v11.csv"
```

```python
import pandas as pd

df = pd.read_csv(data)
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
def clean_text(text: str) -> str:
    text = str(text).strip()
    text = " ".join(text.split())
    return text

df["instruction"] = df["instruction"].apply(clean_text)
df["response"] = df["response"].apply(clean_text)
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates(subset=["instruction"])
df.shape
```

```python
df["category"].value_counts()
```

```python
df["intent"].value_counts().head(10)
```

```python
counts = df["intent"].value_counts()
print(f"all intents number: {df['intent'].nunique()}")
print(f"max intents number: {counts.max()}")
print(f"min intents number: {counts.min()}")
print(f"mean: {counts.mean():.0f}")
```

```python
df["instruction_word_count"] = df["instruction"].str.split().str.len()
df["instruction_word_count"].describe()
```

```python
# فحص عمود flags قبل ما نقفل اليوم
df["flags"].value_counts().head(20)
```

```python
from sklearn.model_selection import train_test_split

train_df, test_df = train_test_split(
    df,
    test_size=0.2,
    random_state=42,
    stratify=df["intent"]
)

train_df.shape, test_df.shape
```

```python
save_path = "/content/drive/MyDrive/TicketMind/data/"

train_df.to_csv(save_path + "train_data.csv", index=False)
test_df.to_csv(save_path + "test_data.csv", index=False)
df.to_csv(save_path + "full_clean_data.csv", index=False)
```
